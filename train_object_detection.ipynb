## 2. Faster R-CNN
Faster R-CNN is a classic two-stage detector. PyTorch's `torchvision` provides a straightforward way to load and modify the classification head.

```python
import torch
import torchvision
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor

def get_faster_rcnn_model(num_classes):
    # Load a model pre-trained on COCO
    model = torchvision.models.detection.fasterrcnn_resnet50_fpn(weights="DEFAULT")
    
    # Get the number of input features for the classifier
    in_features = model.roi_heads.box_predictor.cls_score.in_features
    
    # Replace the pre-trained head with a new one for your custom number of classes
    # Note: num_classes should include the background class (e.g., 1 object class + 1 background = 2)
    model.roi_heads.box_predictor = FastRCNNPredictor(in_features, num_classes)
    
    return model

# Initialize model
NUM_CLASSES_FRCNN = 4 # UPDATE this based on your dataset
frcnn_model = get_faster_rcnn_model(NUM_CLASSES_FRCNN)
frcnn_model.to('cuda' if torch.cuda.is_available() else 'cpu')

# Optimizer setup 
params = [p for p in frcnn_model.parameters() if p.requires_grad]
optimizer_frcnn = torch.optim.SGD(params, lr=0.005, momentum=0.9, weight_decay=0.0005)

# Example logic for training loop (you will need to implement a Dataloader):
# for images, targets in dataloader:
#     images = list(image.to(device) for image in images)
#     targets = [{k: v.to(device) for k, v in t.items()} for t in targets]
#     loss_dict = frcnn_model(images, targets)
#     losses = sum(loss for loss in loss_dict.values())
#     optimizer_frcnn.zero_grad()
#     losses.backward()
#     optimizer_frcnn.step()

print("Faster R-CNN model ready for training!")
```

```python
import os
import json
import torch
from PIL import Image
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms.functional as F
from tqdm import tqdm
import torch.amp

class COCODatasetForFRCNN(Dataset):
    def __init__(self, images_dir, annotation_file):
        self.images_dir = images_dir
        
        print(f"Loading annotations from {annotation_file}...")
        with open(annotation_file, 'r') as f:
            self.coco_data = json.load(f)
            
        self.images = {img['id']: img for img in self.coco_data['images']}
        self.image_ids = list(self.images.keys())
        
        self.annotations = {img_id: [] for img_id in self.image_ids}
        for ann in self.coco_data['annotations']:
            self.annotations[ann['image_id']].append(ann)

    def __len__(self):
        return len(self.image_ids)

    def __getitem__(self, idx):
        img_id = self.image_ids[idx]
        img_info = self.images[img_id]
        
        img_path = os.path.join(self.images_dir, img_info['file_name'])
        img = Image.open(img_path).convert("RGB")
        image_tensor = F.to_tensor(img)

        boxes = []
        labels = []
        area = []
        iscrowd = []
        
        for ann in self.annotations[img_id]:
            x_min, y_min, width, height = ann['bbox']
            boxes.append([x_min, y_min, x_min + width, y_min + height])
            labels.append(ann['category_id']) 
            area.append(ann['area'])
            iscrowd.append(ann.get('iscrowd', 0))

        if len(boxes) > 0:
            boxes = torch.as_tensor(boxes, dtype=torch.float32)
            labels = torch.as_tensor(labels, dtype=torch.int64)
            area = torch.as_tensor(area, dtype=torch.float32)
            iscrowd = torch.as_tensor(iscrowd, dtype=torch.int64)
        else:
            boxes = torch.empty((0, 4), dtype=torch.float32)
            labels = torch.empty((0,), dtype=torch.int64)
            area = torch.empty((0,), dtype=torch.float32)
            iscrowd = torch.empty((0,), dtype=torch.int64)
            
        target = {
            "boxes": boxes,
            "labels": labels,
            "image_id": torch.tensor([img_id]),
            "area": area,
            "iscrowd": iscrowd
        }

        return image_tensor, target

def collate_fn(batch):
    return tuple(zip(*batch))

# ==========================================
# 1. Instantiate the Dataloader
# ==========================================
train_dataset = COCODatasetForFRCNN(
    images_dir=r'd:\Sansarana\Faster R-CNN\train',
    annotation_file=r'd:\Sansarana\Faster R-CNN\train\_annotations.coco.json'
)

# Enable CuDNN benchmark for hardware-level optimizations
torch.backends.cudnn.benchmark = True

train_loader = DataLoader(
    train_dataset, 
    batch_size=2,
    shuffle=True, 
    collate_fn=collate_fn, 
    num_workers=0,
    pin_memory=True
)

# ==========================================
# 2. Start the Training Loop
# ==========================================
device = 'cuda' if torch.cuda.is_available() else 'cpu'

# Move model to device
frcnn_model.to(device)
frcnn_model.train()

print(f"Beginning AMP-Optimized Faster R-CNN Training on {len(train_dataset)} COCO images...")

scaler = torch.amp.GradScaler('cuda')

for epoch in range(15):
    epoch_loss = 0
    progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/15")
    
    for images, targets in progress_bar:
        images = list(img.to(device, non_blocking=True) for img in images)
        targets = [{k: v.to(device, non_blocking=True) for k, v in t.items()} for t in targets]
        
        optimizer_frcnn.zero_grad(set_to_none=True)
        
        with torch.amp.autocast('cuda'):
            loss_dict = frcnn_model(images, targets)
            losses = sum(loss for loss in loss_dict.values())
        
        scaler.scale(losses).backward()
        scaler.step(optimizer_frcnn)
        scaler.update()
        
        epoch_loss += losses.item()
        progress_bar.set_postfix(loss=f"{losses.item():.4f}")
        
    print(f"Epoch {epoch+1} Completed | Average Loss: {epoch_loss/len(train_loader):.4f}")
```

# 3. EfficientDet (VOC Format)
Next, we will train an **EfficientDet** model using the `effdet` package on your Pascal VOC dataset.

```python
!pip install -q effdet omegaconf
import os
import xml.etree.ElementTree as ET
import torch
from PIL import Image
from torch.utils.data import DataLoader, Dataset
import torchvision.transforms.functional as F
from tqdm import tqdm
from effdet.factory import create_model

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# 1. Define Dataset Mapping
CLASS_MAP = {'seated': 1, 'standing': 2, 'reclining': 3}

# 2. PyTorch VOC Dataset for EfficientDet
class EffDetVOCDataset(Dataset):
    def __init__(self, root_dir):
        self.root_dir = root_dir
        self.files = [f.split('.xml')[0] for f in os.listdir(root_dir) if f.endswith('.xml')]
        
    def __len__(self):
        return len(self.files)
        
    def __getitem__(self, idx):
        file_id = self.files[idx]
        img_path = os.path.join(self.root_dir, file_id + '.jpg')
        xml_path = os.path.join(self.root_dir, file_id + '.xml')
        
        img = Image.open(img_path).convert('RGB')
        img_tensor = F.to_tensor(img)
        
        tree = ET.parse(xml_path)
        root = tree.getroot()
        
        boxes = []
        labels = []
        for obj in root.findall('object'):
            name = obj.find('name').text
            if name not in CLASS_MAP:
                continue
            labels.append(CLASS_MAP[name])
            
            bndbox = obj.find('bndbox')
            xmin = float(bndbox.find('xmin').text)
            ymin = float(bndbox.find('ymin').text)
            xmax = float(bndbox.find('xmax').text)
            ymax = float(bndbox.find('ymax').text)
            # EfficientDet natively expects standard (y_min, x_min, y_max, x_max)
            boxes.append([ymin, xmin, ymax, xmax]) 
        
        # Targets must be correctly padded for DataLoader stack
        target = {
            'bbox': torch.tensor(boxes, dtype=torch.float32) if boxes else torch.zeros((0,4)),
            'cls': torch.tensor(labels, dtype=torch.int64)
        }
        return img_tensor, target

def effdet_collate(batch):
    # Pad images natively to multiples of 128 for BiFPN
    padded_images = []
    for b in batch:
        img = b[0]
        h, w = img.shape[-2:]
        pad_h = (128 - (h % 128)) % 128
        pad_w = (128 - (w % 128)) % 128
        padded_images.append(torch.nn.functional.pad(img, (0, pad_w, 0, pad_h), value=0.0))
    images = torch.stack(padded_images)
    targets = [b[1] for b in batch]
    # Convert targets to the format EfficientDet train bench requires:
    # It heavily expects 'bbox' to be shape [batch_size, num_targets, 4]
    # Which means we need to pad the bounding boxes across batch dimension!
    max_targets = max([t['bbox'].shape[0] for t in targets])
    
    padded_bboxes = torch.zeros((len(targets), max_targets, 4))
    padded_classes = torch.zeros((len(targets), max_targets), dtype=torch.int64) - 1  # -1 for empty
    
    for i, t in enumerate(targets):
        m = t['bbox'].shape[0]
        if m > 0:
            padded_bboxes[i, :m] = t['bbox']
            padded_classes[i, :m] = t['cls']
            
    batch_targets = {'bbox': padded_bboxes, 'cls': padded_classes}
    return images, batch_targets

# Instantiate Loader
train_set = EffDetVOCDataset(r'd:\Sansarana\EfficientDetVOC\train')
eff_loader = DataLoader(train_set, batch_size=4, shuffle=True, collate_fn=effdet_collate, num_workers=0)

# 3. Create Model
print("Building EfficientDet...")
effdet_model = create_model('tf_efficientdet_d0', bench_task='train', num_classes=3, pretrained=True, bench_labeler=True)
effdet_model = effdet_model.to(device)

# 4. Training Setup
optimizer = torch.optim.AdamW(effdet_model.parameters(), lr=1e-4)

# 5. Training Loop
effdet_model.train()
num_epochs = 15
scaler = torch.amp.GradScaler('cuda')

print(f"Staring EfficientDet training on {len(train_set)} images...")
for epoch in range(num_epochs):
    epoch_loss = 0.0
    with tqdm(eff_loader, desc=f"Epoch {epoch+1}/{num_epochs}") as pbar:
        for images, targets in pbar:
            images = images.to(device)
            targets = {
                'bbox': targets['bbox'].to(device),
                'cls': targets['cls'].to(device)
            }
            
            optimizer.zero_grad(set_to_none=True)
            
            with torch.amp.autocast('cuda'):
                loss_dict = effdet_model(images, targets)
                loss = loss_dict['loss']
                
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
            
            epoch_loss += loss.item()
            pbar.set_postfix(loss=loss.item())
            
    print(f"Epoch {epoch+1} Average Loss: {epoch_loss/len(eff_loader):.4f}")

torch.save(effdet_model.state_dict(), r'd:\Sansarana\efficientdet_final.pth')
print("EfficientDet training completed and model saved!")
```

### Evaluate EfficientDet Metrics
Just like Faster R-CNN, we evaluate our EfficientDet model using standard **COCO Mean Average Precision (mAP)**. We switch the model to a `bench_task='predict'` wrapper to automatically decode the complex BiFPN features into distinct bounding boxes.

```python
import torchmetrics.utilities.imports
import torchmetrics.detection.helpers
import torchmetrics.detection.mean_ap
torchmetrics.utilities.imports._PYCOCOTOOLS_AVAILABLE = True
torchmetrics.utilities.imports._FASTER_COCO_EVAL_AVAILABLE = True
torchmetrics.detection.helpers._PYCOCOTOOLS_AVAILABLE = True
torchmetrics.detection.helpers._FASTER_COCO_EVAL_AVAILABLE = True
torchmetrics.detection.mean_ap._PYCOCOTOOLS_AVAILABLE = True
torchmetrics.detection.mean_ap._FASTER_COCO_EVAL_AVAILABLE = True
from torchmetrics.detection.mean_ap import MeanAveragePrecision

# 1. Rebuild EfficientDet explicitly for 'predict' task. This wraps the model in a decoder 
# so we get clean [xmin, ymin, xmax, ymax, score, label] boxes instead of raw computational loss graphs.
print("Loading DetBenchPredict for EfficientDet Validation...")
eval_model = create_model('tf_efficientdet_d0', bench_task='predict', num_classes=3, pretrained=False)
eval_model.load_state_dict(torch.load(r'd:\Sansarana\efficientdet_final.pth'))
eval_model = eval_model.to(device)
eval_model.eval()

# 2. Load Validation Set
val_set = EffDetVOCDataset(r'd:\Sansarana\EfficientDetVOC\valid')
val_loader = DataLoader(val_set, batch_size=2, shuffle=False, collate_fn=effdet_collate, num_workers=0)

# 3. Setup COCO mAP metric tracking
metric = MeanAveragePrecision(box_format='xyxy', class_metrics=True, backend='faster_coco_eval')

# 4. Validate
with torch.no_grad():
    for images, batch_targets in tqdm(val_loader, desc="Evaluating EfficientDet"):
        images = images.to(device)
        
        # effdet predict bench natively requires img_scale and img_size metadata for coordinate un-scaling
        # Since our padded size is always 512x512, we explicitly broadcast scale=1.0, size=512x512 
        batch_size = images.shape[0]
        img_info = {
            'img_scale': torch.ones((batch_size,)).to(device),
            'img_size': torch.tensor([[512, 512]] * batch_size, dtype=torch.float).to(device)
        }
        
        with torch.amp.autocast('cuda'):
            # Predict wrapper outputs a dense tensor [Batch, NumBoxes, 6]
            # Structure: [xmin, ymin, xmax, ymax, score, class_id]
            detections = eval_model(images, img_info=img_info)
            
        metric_targets = []
        metric_preds = []
        
        for i in range(batch_size):
            # --- Parse Targets ---
            # Retrieve only non-padded targets
            valid_targets_mask = batch_targets['cls'][i] != -1
            valid_boxes = batch_targets['bbox'][i][valid_targets_mask]
            valid_labels = batch_targets['cls'][i][valid_targets_mask]
            
            # EfficientDet internally tracks targets natively as ymin, xmin, ymax, xmax.
            # But torchmetrics `xyxy` box_format expects xmin, ymin, xmax, ymax.
            # So we must invert [ymin, xmin, ymax, xmax] -> [xmin, ymin, xmax, ymax]
            if valid_boxes.shape[0] > 0:
                valid_boxes = valid_boxes[:, [1, 0, 3, 2]] 
                
            metric_targets.append({
                "boxes": valid_boxes.to(device),
                "labels": valid_labels.to(device)
            })
            
            # --- Parse Predictions ---
            det = detections[i] # [NumBoxes, 6]
            # Filter out extreme noise threshold (e.g. score > 0.05)
            det = det[det[:, 4] > 0.05]
            
            # The bench output is already mapped exactly back as xmin, ymin, xmax, ymax!
            metric_preds.append({
                "boxes": det[:, 0:4].to(device),
                "scores": det[:, 4].to(device),
                "labels": det[:, 5].to(torch.int64).to(device)
            })
            
        metric.update(metric_preds, metric_targets)

results = metric.compute()
print("\n--- EfficientDet Final Metrics ---")
print(f"mAP (IoU 0.50:0.95): {results['map'].item():.4f}")
print(f"mAP@50 (IoU 0.50):   {results['map_50'].item():.4f}")
print(f"mAP@75 (IoU 0.75):   {results['map_75'].item():.4f}")
```

# 4. Grounding DINO (Vision-Language Fine-Tuning)
Grounding DINO detects objects using **text prompts** instead of class IDs. We freeze the heavy Swin & BERT backbones to stay within 12GB VRAM, and fine-tune only the decoder heads.

```python
!pip install -q transformers peft
import os, torch
from PIL import Image
from torch.utils.data import DataLoader, Dataset
from tqdm import tqdm
from transformers import AutoProcessor, AutoModelForZeroShotObjectDetection

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# YOLO class id -> phrase position in prompt
# data.yaml: 0:reclining, 1:seated, 2:standing
TEXT_PROMPT = "reclining statue. seated statue. standing statue."
YOLO_TO_IDX = {0: 0, 1: 1, 2: 2}

# ---- Dataset ----
class YOLODinoDataset(Dataset):
    def __init__(self, images_dir, labels_dir, processor):
        self.images_dir = images_dir
        self.labels_dir = labels_dir
        self.processor = processor
        self.images = [f for f in os.listdir(images_dir) if f.lower().endswith(('.jpg', '.png'))]

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        img_name = self.images[idx]
        base = os.path.splitext(img_name)[0]
        image = Image.open(os.path.join(self.images_dir, img_name)).convert('RGB')
        w, h = image.size

        boxes, class_labels = [], []
        label_path = os.path.join(self.labels_dir, base + '.txt')
        if os.path.exists(label_path):
            with open(label_path) as f:
                for line in f:
                    parts = line.strip().split()
                    if len(parts) < 5:
                        continue
                    cls = int(parts[0])
                    cx, cy, bw, bh = map(float, parts[1:5])
                    xmin = (cx - bw/2) * w
                    ymin = (cy - bh/2) * h
                    xmax = (cx + bw/2) * w
                    ymax = (cy + bh/2) * h
                    if cls in YOLO_TO_IDX:
                        boxes.append([xmin, ymin, xmax, ymax])
                        class_labels.append(YOLO_TO_IDX[cls])

        inputs = self.processor(images=image, text=TEXT_PROMPT, return_tensors='pt')
        inputs = {k: v.squeeze(0) for k, v in inputs.items()}

        target = {
            'boxes': torch.tensor(boxes, dtype=torch.float32) if boxes else torch.zeros((0, 4)),
            'class_labels': torch.tensor(class_labels, dtype=torch.long) if class_labels else torch.zeros((0,), dtype=torch.long)
        }
        return inputs, target

def dino_collate(batch):
    input_ids       = torch.stack([x[0]['input_ids'] for x in batch])
    attention_mask  = torch.stack([x[0]['attention_mask'] for x in batch])
    pixel_values    = torch.stack([x[0]['pixel_values'] for x in batch])
    labels = [x[1] for x in batch]
    return {'input_ids': input_ids, 'attention_mask': attention_mask,
            'pixel_values': pixel_values, 'labels': labels}

# ---- Load Processor & Model ----
MODEL_ID = 'IDEA-Research/grounding-dino-base'
print('Loading Grounding DINO processor...')
processor = AutoProcessor.from_pretrained(MODEL_ID)

train_set = YOLODinoDataset(
    r'd:\Sansarana\RT-DETR\train\images',
    r'd:\Sansarana\RT-DETR\train\labels',
    processor
)
dino_loader = DataLoader(train_set, batch_size=2, shuffle=True,
                         collate_fn=dino_collate, num_workers=0)

print('Loading Grounding DINO model weights...')
dino_model = AutoModelForZeroShotObjectDetection.from_pretrained(
    MODEL_ID, ignore_mismatched_sizes=True
)

# Freeze vision + language backbones; only train decoder/prediction heads
# HF structure: dino_model.model.backbone (Swin) + dino_model.model.text_backbone (BERT)
for param in dino_model.model.backbone.parameters():
    param.requires_grad = False
for param in dino_model.model.text_backbone.parameters():
    param.requires_grad = False

trainable = sum(p.numel() for p in dino_model.parameters() if p.requires_grad)
total     = sum(p.numel() for p in dino_model.parameters())
print(f'Trainable params: {trainable/1e6:.1f}M / {total/1e6:.1f}M total')

dino_model = dino_model.to(device)
optimizer  = torch.optim.AdamW(
    filter(lambda p: p.requires_grad, dino_model.parameters()), lr=1e-5
)
scaler = torch.amp.GradScaler('cuda')

# ---- Training Loop ----
num_epochs = 5
dino_model.train()
print(f'Starting Grounding DINO fine-tuning on {len(train_set)} images...')

for epoch in range(num_epochs):
    epoch_loss = 0.0
    with tqdm(dino_loader, desc=f'Epoch {epoch+1}/{num_epochs}') as pbar:
        for batch in pbar:
            pixel_values   = batch['pixel_values'].to(device)
            input_ids      = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            labels = [{k: v.to(device) for k, v in t.items()} for t in batch['labels']]

            optimizer.zero_grad(set_to_none=True)

            with torch.amp.autocast('cuda'):
                outputs = dino_model(
                    pixel_values=pixel_values,
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    labels=labels
                )
                loss = outputs.loss

            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(dino_model.parameters(), max_norm=1.0)
            scaler.step(optimizer)
            scaler.update()

            epoch_loss += loss.item()
            pbar.set_postfix(loss=f'{loss.item():.4f}')

    print(f'Epoch {epoch+1} Average Loss: {epoch_loss/len(dino_loader):.4f}')

torch.save(dino_model.state_dict(), r'd:\Sansarana\dino_tuned_final.pth')
print('Grounding DINO fine-tuning complete! Saved to dino_tuned_final.pth')
```
